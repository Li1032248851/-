```python
def caesar_cipher(text, key, alphabet):
    """
    实现凯撒密码的加密与解密
    Реализация шифра Цезаря для шифрования и дешифрования
    """
    result = ""
    m = len(alphabet) # 字母表长度 / Длина алфавита

    for char in text:
        if char.lower() in alphabet:
            # 找到字母在字母表中的位置
            # Находим позицию буквы в алфавите
            idx = alphabet.find(char.lower())

            # 计算位移后的新位置
            # Вычисляем новую позицию со сдвигом
            new_idx = (idx + key) % m

            # 保持大小写一致
            # Сохраняем регистр
            new_char = alphabet[new_idx]
            result += new_char.upper() if char.isupper() else new_char
        else:
            # 非字母字符直接保留（如空格、标点）
            # Неалфавитные символы остаются без изменений
            result += char
    return result
```

```python
def atbash_cipher(text, alphabet):
    """
    实现阿特巴什密码（镜像替换）
    Реализация шифра Атбаш (зеркальная замена)
    """
    result = ""
    # 创建镜像字母表（将原字母表反转）
    # Создаем зеркальный алфавит (переворачиваем оригинал)
    reversed_alphabet = alphabet[::-1]

    for char in text:
        if char.lower() in alphabet:
            idx = alphabet.find(char.lower())
            # 使用反转后的字母表进行替换
            # Заменяем букву на соответствующую из перевернутого алфавита
            new_char = reversed_alphabet[idx]
            result += new_char.upper() if char.isupper() else new_char
        else:
            result += char
    return result
```

```python
# --- 测试部分 / Тестирование ---

# 定义俄语字母表（包含文档中提到的33个字母）
# Определение русского алфавита (33 буквы)
ru_alphabet = "абвгдеёжзийклмнопрстуфхцчшщъыьэюя"

# 1. 凯撒密码测试 (密钥 k=3)
# Тест шифра Цезаря (ключ k=3)
original_text = "Пришел увидел победил"
caesar_encrypted = caesar_cipher(original_text, 3, ru_alphabet)
caesar_decrypted = caesar_cipher(caesar_encrypted, -3, ru_alphabet)

print(f"--- Caesar Cipher (k=3) ---")
print(f"Original:  {original_text}")
print(f"Encrypted: {caesar_encrypted}")
print(f"Decrypted: {caesar_decrypted}\n")

# 2. 阿特巴什密码测试
# Тест шифра Атбаш
atbash_text = "Саратов"
atbash_encrypted = atbash_cipher(atbash_text, ru_alphabet)

print(f"--- Atbash Cipher ---")
print(f"Original:  {atbash_text}")
print(f"Encrypted: {atbash_encrypted}")
```
