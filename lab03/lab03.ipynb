```python
def gamma_cipher(text, key, mode='encrypt'):
    # 中文：定义俄语字母表（33个字母）
    # RU: Определяем русский алфавит (33 буквы)
    alphabet = "АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ"
    N = len(alphabet)

    text = text.upper()
    key = key.upper()
    result = ""

    # 中文：遍历明文中的每一个字符
    # RU: Проходим через каждый символ исходного текста
    for i in range(len(text)):
        if text[i] in alphabet:
            # 中文：获取当前字符和密钥字符在字母表中的索引位置
            # RU: Получаем индексы текущего символа и символа гаммы в алфавите
            p_idx = alphabet.index(text[i])
            # 中文：使用取模运算让密钥(Gamma)循环使用
            # RU: Используем операцию остатка от деления для зацикливания гаммы
            k_idx = alphabet.index(key[i % len(key)])

            if mode == 'encrypt':
                # 中文：加密公式：(明文索引 + 密钥索引) 模 N
                # RU: Формула шифрования: (индекс текста + индекс ключа) mod N
                res_idx = (p_idx + k_idx) % N
            else:
                # 中文：解密公式：(密文索引 - 密钥索引 + N) 模 N
                # RU: Формула дешифрования: (индекс шифра - индекс ключа + N) mod N
                res_idx = (p_idx - k_idx + N) % N

            result += alphabet[res_idx]
        else:
            # 中文：如果字符不在字母表中（如空格），原样保留
            # RU: Если символа нет в алфавите (например, пробел), оставляем как есть
            result += text[i]

    return result

# 示例运行 / Пример запуска
plaintext = "ПРИКАЗ"
gamma = "ГАММА"
encrypted = gamma_cipher(plaintext, gamma, mode='encrypt')
print(f"加密结果: {encrypted}")
```
