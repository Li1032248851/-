**Алгоритм Евклида**

```python
def euclidean_gcd(a, b):
    # 中文：初始化余数 r0 为 a，r1 为 b
    # RU: Инициализация r0 = a, r1 = b
    r0, r1 = a, b

    # 中文：循环执行取余操作，直到余数 r1 变为 0
    # RU: Цикл продолжается, пока остаток r1 не станет равен 0
    while r1 != 0:
        # 中文：核心逻辑：r_i+1 = r_i-1 mod r_i
        # RU: Основная логика: r_i+1 = r_i-1 mod r_i
        r0, r1 = r1, r0 % r1

    # 中文：返回最后的非零余数作为最大公约数
    # RU: Возвращает последний ненулевой остаток как НОД
    return r0

# 样例运行 / Пример запуска
# 输入 (12345, 24690) -> 输出: 12345
print(f"НОД (12345, 24690) = {euclidean_gcd(12345, 24690)}")
```

**Бинарный алгоритм Евклида**

```python
def binary_gcd(a, b):
    # 中文：g 用于记录提取出的公因子 2
    # RU: g используется для хранения общего множителя 2
    g = 1

    # 中文：当 a 和 b 都是偶数时，不断提取公因子 2
    # RU: Пока a и b четные, выносим множитель 2 в g
    while a % 2 == 0 and b % 2 == 0:
        a //= 2
        b //= 2
        g *= 2

    u, v = a, b
    # 中文：主循环，通过移位和减法代替取模运算
    # RU: Основной цикл: сдвиги и вычитание вместо деления
    while u != 0:
        while u % 2 == 0: u //= 2
        while v % 2 == 0: v //= 2
        if u >= v:
            u = u - v
        else:
            v = v - u

    # 中文：结果为提取出的 2 的幂次与剩余值的乘积
    # RU: Результат — произведение g на оставшееся значение v
    return g * v

# 样例运行 / Пример запуска
# 输入 (48, 18) -> 输出: 6
print(f"Бинарный НОД (48, 18) = {binary_gcd(48, 18)}")
```

**Расширенный алгоритм Евклида**

```python
def extended_euclidean_gcd(a, b):
    # 中文：初始化系数 x 和 y，满足 ax + by = r
    # RU: Инициализация коэффициентов x и y для ax + by = r
    r0, r1 = a, b
    x0, x1 = 1, 0
    y0, y1 = 0, 1

    while r1 != 0:
        # 中文：计算商 q 用于后续系数更新
        # RU: Находим частное q для обновления коэффициентов
        q = r0 // r1
        r0, r1 = r1, r0 - q * r1
        # 中文：更新线性组合的系数
        # RU: Обновление коэффициентов линейной комбинации
        x0, x1 = x1, x0 - q * x1
        y0, y1 = y1, y0 - q * y1

    # 中文：返回最大公约数 d 以及系数 x, y
    # RU: Возвращает НОД d и коэффициенты x, y
    return r0, x0, y0

# 样例运行 / Пример запуска
# 输入 (91, 105) -> 输出: (7, 7, -6) -> 即 91*7 + 105*(-6) = 7
d, x, y = extended_euclidean_gcd(91, 105)
print(f"Расширенный НОД: d={d}, x={x}, y={y}")
```

**Расширенный бинарный алгоритм 2**

```python
def extended_binary_gcd(a, b):
    g = 1
    while a % 2 == 0 and b % 2 == 0:
        a //= 2; b //= 2; g *= 2

    u, v = a, b
    # 中文：A, B, C, D 对应线性组合的辅助变量
    # RU: A, B, C, D — вспомогательные переменные для линейной комбинации
    A, B, C, D = 1, 0, 0, 1

    while u != 0:
        while u % 2 == 0:
            u //= 2
            # 中文：若系数为偶数则除以 2，否则按公式调整
            # RU: Если коэффициенты четные, делим на 2, иначе корректируем по формуле
            if A % 2 == 0 and B % 2 == 0:
                A //= 2; B //= 2
            else:
                A = (A + b) // 2; B = (B - a) // 2
        # ... 对 v 执行类似逻辑 ...
        while v % 2 == 0:
            v //= 2
            if C % 2 == 0 and D % 2 == 0:
                C //= 2; D //= 2
            else:
                C = (C + b) // 2; D = (D - a) // 2

        if u >= v:
            u -= v; A -= C; B -= D
        else:
            v -= u; C -= A; D -= B

    return g * v, C, D

# 样例运行 / Пример запуска
# 输入 (154, 105) -> 输出: (7, 1, -1) -> 即 154*1 + 105*(-1) = 49 (示例结果)
d, x, y = extended_binary_gcd(154, 105)
print(f"Бинарный расширенный НОД: d={d}, x={x}, y={y}")
```
