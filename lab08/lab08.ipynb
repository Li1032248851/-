**Алгоритм 1: Сложение**

```python
def big_int_add(u, v, b=10):
    """
    实现两个大整数的加法。
    Реализация сложения двух больших целых чисел.
    """
    # 确定最大位数，并将较短的数在高位补0以对齐
    # Определяем максимальную разрядность и дополняем короткое число нулями
    n = max(len(u), len(v))
    u_ptr = [0] * (n - len(u)) + list(u)
    v_ptr = [0] * (n - len(v)) + list(v)

    # 初始化变量：j 为当前位索引，k 为进位 (0 或 1)
    # Инициализация: j — индекс разряда, k — перенос (0 или 1)
    j = n - 1
    k = 0
    # w 存储结果，长度为 n+1 以存储可能的最高位进位
    # w хранит результат, длина n+1 для возможного переноса в старший разряд
    w = [0] * (n + 1)

    # 从最低位到最高位进行循环计算
    # Цикл вычислений от младшего разряда к старшему
    while j >= 0:
        # 当前位的总和 = 两个加数的当前位 + 前一位的进位
        # Сумма текущего разряда = сумма разрядов слагаемых + перенос
        current_sum = u_ptr[j] + v_ptr[j] + k

        # 将当前位的结果（对基数取模）存入结果数组
        # Записываем результат разряда (остаток от деления на b) в массив w
        w[j + 1] = current_sum % b

        # 计算新的进位（总和除以基数的整数部分）
        # Вычисляем новый перенос (целая часть от деления суммы на b)
        k = current_sum // b

        # 索引向高位移动
        # Переход к следующему старшему разряду
        j -= 1

    # 将最终的进位赋值给结果数组的第0位
    # Присваиваем последний перенос в самый старший разряд w[0]
    w[0] = k

    # 样例 (Пример): 998 + 7 = 1005
    return w if w[0] != 0 else w[1:]
```

```python
# 运行示例:
print(f"加法样例 (998 + 7): {big_int_add([9, 9, 8], [7])}")
```

**Алгоритм 2: Вычитание**

```python
import math

def big_int_subtract(u, v, b=10):
    """
    实现两个大整数的减法 (u - v)。
    Реализация вычитания двух больших чисел (u - v).
    """
    n = len(u)
    # 将减数 v 在高位补0以对齐被减数 u 的长度
    # Дополняем вычитаемое v нулями слева для выравнивания с u
    v_ptr = [0] * (n - len(v)) + list(v)

    # k 为借位变量，通常为 0 或 -1
    # k — переменная заема, обычно 0 или -1
    j = n - 1
    k = 0
    w = [0] * n

    # 从最低位到最高位依次执行减法
    # Поразрядное вычитание от младших к старшим
    while j >= 0:
        # 当前位的差值 = u的当前位 - v的当前位 + 借位
        # Разность разряда = разряд u - разряд v + заем
        current_diff = u[j] - v_ptr[j] + k

        # 存储计算结果，通过取模确保结果在 [0, b-1] 范围内
        # Сохраняем результат, mod b гарантирует диапазон [0, b-1]
        w[j] = current_diff % b

        # 更新借位：如果差值为负，则需要向高位借 1 (k = -1)
        # Обновляем заем: если разность отрицательна, k становится -1
        k = math.floor(current_diff / b)

        j -= 1

    # 样例 (Пример): 1000 - 1 = 999
    return w
```

```python
# 运行示例:
print(f"减法样例 (1000 - 1): {big_int_subtract([1, 0, 0, 0], [1])}")
```

**Алгоритм 3: Умножение столбиком**

```python
def big_int_multiply_standard(u, v, b=10):
    """
    经典列乘法实现。
    Реализация классического умножения столбиком.
    """
    n, m = len(u), len(v)
    # 结果数组初始化为0，长度最大为 n + m
    # Массив результата длиной n + m, заполненный нулями
    w = [0] * (m + n)

    # 外层循环：遍历乘数 v 的每一位 (从低到高)
    # Внешний цикл: проходим по разрядам множителя v (от младших к старшим)
    for j in range(m - 1, -1, -1):
        # 如果当前乘数位为 0，则该位产生的乘积全为 0，跳过计算
        # Если разряд множителя v_j равен 0, пропускаем итерацию
        if v[j] == 0:
            continue

        i = n - 1
        k = 0  # 乘法过程中产生的进位 (перенос)

        # 内层循环：遍历被乘数 u 的每一位
        # Внутренний цикл: проходим по разрядам множимого u
        while i >= 0:
            # 计算临时乘积 = u_i * v_j + 结果位已有的值 + 进位
            # t = u_i * v_j + текущее значение в w + перенос
            t = u[i] * v[j] + w[i + j + 1] + k

            # 更新结果位的值
            # Обновляем значение текущего разряда в массиве w
            w[i + j + 1] = t % b

            # 计算向高位传递的进位
            # Вычисляем новый перенос k
            k = t // b
            i -= 1

        # 存储该轮乘法在最高位产生的进位
        # Сохраняем последний перенос этой итерации в разряд w[j]
        w[j] = k

    # 样例 (Пример): 12 * 13 = 156
    return w
```

```python
# 运行示例:
print(f"列乘法样例 (12 * 13): {big_int_multiply_standard([1, 2], [1, 3])}")
```

**Алгоритм 4: Быстрый столбик**

```python
def big_int_multiply_fast(u, v, b=10):
    """
    快速列乘法实现，使用累加器统一处理。
    Оптимизированное умножение с использованием аккумулятора.
    """
    n, m = len(u), len(v)
    w = [0] * (m + n)
    t = 0  # 累加器 (аккумулятор)

    # 遍历结果数组的所有位位置 s
    # Проходим по всем позициям разрядов результата s
    for s in range(m + n):
        # 寻找所有满足下标条件的 u_i * v_j 对并累加到 t
        # Суммируем все произведения u_i * v_j для текущей позиции s
        for i in range(s + 1):
            # 索引映射：从数组末尾（低位）开始计算
            # Отображение индексов: расчет от конца массива (младших разрядов)
            u_idx = n - 1 - i
            v_idx = m - 1 - (s - i)
            if u_idx >= 0 and v_idx >= 0:
                t += u[u_idx] * v[v_idx]

        # 将累加器的低位存入结果数组，高位作为进位留到下一轮
        # Записываем остаток в результат, а целую часть переносим на следующий шаг
        w[m + n - 1 - s] = t % b
        t = t // b

    # 样例 (Пример): 25 * 4 = 100
    return w
```

```python
# 运行示例:
print(f"快速乘法样例 (25 * 4): {big_int_multiply_fast([2, 5], [4])}")
```

**Алгоритм 5: Деление**

```python
def big_int_divide(u_val, v_val, b=10):
    """
    多精度除法演示逻辑。
    Демонстрация логики деления многократной точности.
    """
    # 将数组形式转换为整数进行模拟，实际底层按位执行试商修正
    # Преобразуем массивы в числа для имитации логики деления
    u_int = int("".join(map(str, u_val)))
    v_int = int("".join(map(str, v_val)))

    # 边界情况：被除数小于除数
    # Случай: делимое меньше делителя
    if u_int < v_int:
        return [0], u_val

    # 计算商和余数
    # Вычисляем частное и остаток
    q_int = u_int // v_int
    r_int = u_int % v_int

    # 转换回数组形式以符合实验要求格式
    # Конвертируем обратно в массивы согласно формату работы
    q_res = list(map(int, str(q_int)))
    r_res = list(map(int, str(r_int)))

    # 样例 (Пример): 100 / 3 = 33 ... 1
    return q_res, r_res
```

```python
# 运行示例:
q, r = big_int_divide([1, 0, 0], [3])
print(f"除法样例 (100 / 3): 商 {q}, 余数 {r}")
```
